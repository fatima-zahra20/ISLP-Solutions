Exercise 2:


We wish to find the class $k$ that maximises $p_k(x)$. Since the denominator of (4.17) does not depend on $k$ it sums over all classes and is identical for every candidate maximising $p_k(x)$ is equivalent to maximising the numerator alone:

$$\pi_k \frac{1}{\sqrt{2\pi}\sigma} \exp\left(-\frac{1}{2\sigma^2}(x-\mu_k)^2\right)$$

The factor $\frac{1}{\sqrt{2\pi}\sigma}$ also carries no $k$ subscript, so it is the same constant across all classes and cancels. We are left with maximising:

$$\pi_k \exp\left(-\frac{1}{2\sigma^2}(x-\mu_k)^2\right)$$

Since $\log$ is a strictly increasing function, the class that maximises an expression also maximises its logarithm. Applying $\log$:

$$\log(\pi_k) - \frac{1}{2\sigma^2}(x-\mu_k)^2$$

Expanding $(x - \mu_k)^2 = x^2 - 2x\mu_k + \mu_k^2$:

$$\log(\pi_k) - \frac{x^2}{2\sigma^2} + \frac{x\mu_k}{\sigma^2} - \frac{\mu_k^2}{2\sigma^2}$$

The term $\frac{x^2}{2\sigma^2}$ carries no $k$ subscript and is therefore identical across all classes. Dropping it leaves:

$$x \cdot \frac{\mu_k}{\sigma^2} - \frac{\mu_k^2}{2\sigma^2} + \log(\pi_k) = \delta_k(x)$$

This is exactly (4.18). Since each cancellation step preserved the ordering across classes, the class maximising $p_k(x)$ is the same as the class maximising $\delta_k(x)$.
