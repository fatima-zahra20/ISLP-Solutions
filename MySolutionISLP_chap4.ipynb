Exercise 2:


We wish to find the class $k$ that maximises $p_k(x)$. Since the denominator of (4.17) does not depend on $k$ it sums over all classes and is identical for every candidate maximising $p_k(x)$ is equivalent to maximising the numerator alone:

$$\pi_k \frac{1}{\sqrt{2\pi}\sigma} \exp\left(-\frac{1}{2\sigma^2}(x-\mu_k)^2\right)$$

The factor $\frac{1}{\sqrt{2\pi}\sigma}$ also carries no $k$ subscript, so it is the same constant across all classes and cancels. We are left with maximising:

$$\pi_k \exp\left(-\frac{1}{2\sigma^2}(x-\mu_k)^2\right)$$

Since $\log$ is a strictly increasing function, the class that maximises an expression also maximises its logarithm. Applying $\log$:

$$\log(\pi_k) - \frac{1}{2\sigma^2}(x-\mu_k)^2$$

Expanding $(x - \mu_k)^2 = x^2 - 2x\mu_k + \mu_k^2$:

$$\log(\pi_k) - \frac{x^2}{2\sigma^2} + \frac{x\mu_k}{\sigma^2} - \frac{\mu_k^2}{2\sigma^2}$$

The term $\frac{x^2}{2\sigma^2}$ carries no $k$ subscript and is therefore identical across all classes. Dropping it leaves:

$$x \cdot \frac{\mu_k}{\sigma^2} - \frac{\mu_k^2}{2\sigma^2} + \log(\pi_k) = \delta_k(x)$$

This is exactly (4.18). Since each cancellation step preserved the ordering across classes, the class maximising $p_k(x)$ is the same as the class maximising $\delta_k(x)$.

Exercise 4(d):


1. **Exponential Shrinkage of Neighborhood Volume:**
* For $p = 1$, a $10\%$ window along the feature axis captures **$10\%$** ($0.10^1$) of the observations.


* For $p = 2$, a $10\%$ window along each axis captures **$1\%$** ($0.10^2$) of the observations.


* For $p = 100$, a $10\%$ window along each axis captures **$10^{-100}$** ($0.10^{100}$) of the observations.




2. **The Curse of Dimensionality in KNN:**
* As P increases, the fraction of available training data contained within any truly local neighborhood shrinks exponentially toward zero ($0.10^p \to 0$).


* For any finite dataset size $N$, the expected number of local neighbors $K = 0.10^p \times N$ becomes effectively **$0$**.


Exercise 4(e):

1. **For $p = 1$ dimension:**

$$s = (0.10)^{1/1} = \mathbf{0.10} \quad \text{(10\% of the feature range)}$$

2. **For $p = 2$ dimensions:**

$$s = (0.10)^{1/2} = \sqrt{0.10} \approx \mathbf{0.316} \quad \text{(31.6\% of each feature's range)}$$

3. **For $p = 100$ dimensions:**

$$s = (0.10)^{1/100} = (0.10)^{0.01} \approx \mathbf{0.977} \quad \text{(97.7\% of each feature's range)}$$



* **Low Dimensions ($p = 1$):** To capture $10\%$ of the data, the window only needs to span **$10\%$** of the single feature's range.


* **Moderate Dimensions ($p = 2$):** To capture $10\%$ of the data (area $= 0.10$), the square box must span **$31.6\%$** of the length along both axes ($0.316 \times 0.316 = 0.10$).


* **High Dimensions ($p = 100$):** To capture that same $10\%$ of the data, the hypercube must cover **$97.7\%$** of the feature range along **every single one of the 100 dimensions**.



### **Conclusion:**

In high-dimensional spaces, **truly local neighborhoods do not exist**. To gather even a tiny fraction ($10\%$) of nearby data points for a prediction, the search window must stretch across almost the entire domain ($97.7\%$) of every feature, defeating the entire purpose of K-Nearest Neighbors.

**1. Training Set Performance**: *QDA performs better*
**Reasoning**: QDA is a **more flexible classifier** than LDA because it estimates separate covariance matrices ($\boldsymbol{\Sigma}_k$) for each class rather than assuming a shared covariance matrix.

**Impact**: Due to its higher flexibility (lower bias, higher variance), QDA can bend its decision boundary to fit sample-specific noise and random variations in the training data.

**Conclusion**: QDA will almost always achieve a **lower training error rate** than LDA.

**2. Test Set Performance:** *LDA performs better*

**Reasoning**: Since the **true Bayes decision boundary is linear**, LDA’s underlying model assumption ($\boldsymbol{\Sigma}_1 = \dots = \boldsymbol{\Sigma}_K = \boldsymbol{\Sigma}$) correctly matches the true data-generating process.

**Impact**: LDA maintains lower variance while remaining unbiased with respect to the true linear boundary. QDA, on the other hand, suffers from unnecessary complexity and higher variance—it will overfit the training set by attempting to fit curved boundaries where none actually exist.

**Conclusion**: LDA will generalize better to unseen data and achieve a **lower test error rate** than QDA.

* Once variance is no longer an issue (because $n$ is large), QDA's lower bias / higher flexibility becomes an unmitigated advantage.


* If the true Bayes boundary is non-linear, QDA will drastically outperform LDA.


* If the true Bayes boundary is linear, QDA's estimated covariance matrices will converge to equality ($\hat{\boldsymbol{\Sigma}}_k \to \boldsymbol{\Sigma}$), making its decision boundary effectively linear and its test accuracy catch up to LDA.

**Justification & Bias-Variance Trade-off**

**Bias Consideration**:
Because the true Bayes decision boundary is **linear**, LDA is unbiased relative to the true boundary.
Although QDA is flexible enough to model a linear boundary, its extra flexibility yields **no reduction in bias**.




**Variance Consideration**:
* QDA estimates a distinct covariance matrix ($\boldsymbol{\Sigma}_k$) for each of the $K$ classes. In contrast, LDA estimates a single shared covariance matrix ($\boldsymbol{\Sigma}$).
Estimating these additional parameters requires estimating significantly more parameters, which introduces **higher model variance**.



**Test Error Rate**:
QDA will tend to overfit the training data by fitting sample-specific noise (trying to draw quadratic curves where the true boundary is linear).
Due to this higher variance without any gain in bias reduction, QDA will lead to a **higher (inferior) test error rate** compared to LDA.
