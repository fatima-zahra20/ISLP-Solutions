We wish to find the class $k$ that maximises $p_k(x)$. Since the denominator of (4.17) does not depend on $k$ it sums over all classes and is identical for every candidate maximising $p_k(x)$ is equivalent to maximising the numerator alone:

$$\pi_k \frac{1}{\sqrt{2\pi}\sigma} \exp\left(-\frac{1}{2\sigma^2}(x-\mu_k)^2\right)$$

The factor $\frac{1}{\sqrt{2\pi}\sigma}$ also carries no $k$ subscript, so it is the same constant across all classes and cancels. We are left with maximising:

$$\pi_k \exp\left(-\frac{1}{2\sigma^2}(x-\mu_k)^2\right)$$

Since $\log$ is a strictly increasing function, the class that maximises an expression also maximises its logarithm. Applying $\log$:

$$\log(\pi_k) - \frac{1}{2\sigma^2}(x-\mu_k)^2$$

Expanding $(x - \mu_k)^2 = x^2 - 2x\mu_k + \mu_k^2$:

$$\log(\pi_k) - \frac{x^2}{2\sigma^2} + \frac{x\mu_k}{\sigma^2} - \frac{\mu_k^2}{2\sigma^2}$$

The term $\frac{x^2}{2\sigma^2}$ carries no $k$ subscript and is therefore identical across all classes. Dropping it leaves:

$$x \cdot \frac{\mu_k}{\sigma^2} - \frac{\mu_k^2}{2\sigma^2} + \log(\pi_k) = \delta_k(x)$$

This is exactly (4.18). Since each cancellation step preserved the ordering across classes, the class maximising $p_k(x)$ is the same as the class maximising $\delta_k(x)$.

* As P increases, the fraction of available training data contained within any truly local neighborhood shrinks exponentially toward zero ($0.10^p \to 0$).

$$s = (0.10)^{1/2} = \sqrt{0.10} \approx \mathbf{0.316} \quad \text{(31.6\% of each feature's range)}$$

* **Low Dimensions ($p = 1$):** To capture $10\%$ of the data, the window only needs to span **$10\%$** of the single feature's range.


* **Moderate Dimensions ($p = 2$):** To capture $10\%$ of the data (area $= 0.10$), the square box must span **$31.6\%$** of the length along both axes ($0.316 \times 0.316 = 0.10$).


* **High Dimensions ($p = 100$):** To capture that same $10\%$ of the data, the hypercube must cover **$97.7\%$** of the feature range along **every single one of the 100 dimensions**.



### **Conclusion:**

In high-dimensional spaces, **truly local neighborhoods do not exist**. To gather even a tiny fraction ($10\%$) of nearby data points for a prediction, the search window must stretch across almost the entire domain ($97.7\%$) of every feature, defeating the entire purpose of K-Nearest Neighbors.

**Conclusion**: LDA will generalize better to unseen data and achieve a **lower test error rate** than QDA.

**1. Identify Given Parameters**

From the problem statement:

* **Classes ($Y$)**:
$\text{Yes}$ (Issued dividend)
//

* **Priors ($\pi_k = P(Y = k)$)**:
$P(Y = \text{Yes}) = \pi_{\text{Yes}} = 0.80$
//

* **Means ($\mu_k$)**:
$\mu_{\text{Yes}} = 10$
//

* **Shared Variance ($\sigma^2$)**:
$\hat{\sigma}^2 = 36 \implies \sigma = 6$


* **Test Value ($x$)**:
$X = 4$

**2. Bayes' Theorem Formula for Normal Distributions**

The probability $P(Y = \text{Yes} \mid X = x)$ is given by:

$$P(Y = \text{Yes} \mid X = x) = \frac{\pi_{\text{Yes}} f_{\text{Yes}}(x)}{\pi_{\text{Yes}} f_{\text{Yes}}(x) + \pi_{\text{No}} f_{\text{No}}(x)}$$

Where $f_k(x)$ is the normal density function:


$$f_k(x) = \frac{1}{\sqrt{2\pi\sigma^2}} \exp\left( -\frac{(x - \mu_k)^2}{2\sigma^2} \right)$$

**3. Compute Normal Densities ($f_k(x)$) for $X = 4$**

**A. For $Y = \text{Yes}$ ($\mu_{\text{Yes}} = 10, \sigma^2 = 36$):**

$$f_{\text{Yes}}(4) = \frac{1}{\sqrt{2\pi \cdot 36}} \exp\left( -\frac{(4 - 10)^2}{2 \cdot 36} \right)$$

$$f_{\text{Yes}}(4) = \frac{1}{\sqrt{72\pi}} \exp\left( -\frac{(-6)^2}{72} \right) = \frac{1}{\sqrt{72\pi}} \exp\left( -\frac{36}{72} \right) = \frac{1}{\sqrt{72\pi}} e^{-0.5}$$

#### **B. For $Y = \text{No}$ ($\mu_{\text{No}} = 0, \sigma^2 = 36$):**

$$f_{\text{No}}(4) = \frac{1}{\sqrt{2\pi \cdot 36}} \exp\left( -\frac{(4 - 0)^2}{2 \cdot 36} \right)$$

$$f_{\text{No}}(4) = \frac{1}{\sqrt{72\pi}} \exp\left( -\frac{16}{72} \right) = \frac{1}{\sqrt{72\pi}} e^{-4/9}$$

**4. Plug into Bayes' Formula & Simplify**

the constant term $\frac{1}{\sqrt{72\pi}}$ cancels out in both the numerator and denominator:

$$P(Y = \text{Yes} \mid X = 4) = \frac{0.80 \cdot e^{-0.5}}{0.80 \cdot e^{-0.5} + 0.20 \cdot e^{-4/9}}$$

**5. Numerical Calculation**

1. **Evaluate Exponentials**:
* $e^{-0.5} \approx 0.60653$
* $e^{-4/9} = e^{-0.4444} \approx 0.64118$


2. **Compute Numerator**:

$$\text{Numerator} = 0.80 \cdot 0.60653 = 0.48522$$


3. **Compute Denominator Terms**:
* $0.80 \cdot e^{-0.5} = 0.48522$
* $0.20 \cdot e^{-4/9} = 0.20 \cdot 0.64118 = 0.12824$
* $\text{Denominator} = 0.48522 + 0.12824 = 0.61346$


4. **Final Probability**:

$$P(Y = \text{Yes} \mid X = 4) = \frac{0.48522}{0.61346} \approx 0.75185$$

**Final Answer**

The predicted probability that the company will issue a dividend given $X = 4$ last year is approximately **$0.752$** (or **$75.2\%$**).

$$\text{log odds} = \log\left(\frac{\exp(\hat{\alpha}_{\text{orange}0} + \hat{\alpha}_{\text{orange}1} x)}{\exp(\hat{\alpha}_{\text{apple}0} + \hat{\alpha}_{\text{apple}1} x)}\right) = (\hat{\alpha}_{\text{orange}0} - \hat{\alpha}_{\text{apple}0}) + (\hat{\alpha}_{\text{orange}1} - \hat{\alpha}_{\text{apple}1}) x$$

(c)

**1. Equating the Log Odds**

Since both models are fit to the same data, their predicted log-odds ratios of $\text{orange}$ versus $\text{apple}$ must be equal:

$$\text{Log Odds} = (\hat{\alpha}_{\text{orange}0} - \hat{\alpha}_{\text{apple}0}) + (\hat{\alpha}_{\text{orange}1} - \hat{\alpha}_{\text{apple}1})x = \hat{\beta}_0 + \hat{\beta}_1 x$$

Given that $\hat{\beta}_0 = 2$ and $\hat{\beta}_1 = -1$:

$$(\hat{\alpha}_{\text{orange}0} - \hat{\alpha}_{\text{apple}0}) + (\hat{\alpha}_{\text{orange}1} - \hat{\alpha}_{\text{apple}1})x = 2 - 1x$$

Matching terms line by line:

* **Intercept constraint:** $\hat{\alpha}_{\text{orange}0} - \hat{\alpha}_{\text{apple}0} = 2$
* **Slope constraint:** $\hat{\alpha}_{\text{orange}1} - \hat{\alpha}_{\text{apple}1} = -1$


**2. General Solution with Constants $c$ and $d$**

Because we have four unknown parameters $(\hat{\alpha}_{\text{orange}0}, \hat{\alpha}_{\text{orange}1}, \hat{\alpha}_{\text{apple}0}, \hat{\alpha}_{\text{apple}1})$ but only two equations, there are infinitely many valid solutions.

We can write the complete family of solutions by introducing arbitrary real constants $c$ for the intercept shift and $d$ for the slope shift:

$$\begin{aligned} \hat{\alpha}_{\text{apple}0} &= c, &\quad \hat{\alpha}_{\text{orange}0} &= 2 + c \\ \hat{\alpha}_{\text{apple}1} &= d, &\quad \hat{\alpha}_{\text{orange}1} &= -1 + d \end{aligned}$$

**3. Proof That $c$ and $d$ Do Not Change the Probability**

To see why $c$ and $d$ do not cancel out or alter the predicted probability, substitute these parameter expressions directly into the softmax probability formula:

$$\begin{aligned} \widehat{\operatorname{Pr}}(Y = \text{orange} \mid X = x) &= \frac{\exp\left(\hat{\alpha}_{\text{orange}0} + \hat{\alpha}_{\text{orange}1}x\right)}{\exp\left(\hat{\alpha}_{\text{orange}0} + \hat{\alpha}_{\text{orange}1}x\right) + \exp\left(\hat{\alpha}_{\text{apple}0} + \hat{\alpha}_{\text{apple}1}x\right)} \\ &= \frac{\exp\left((2 + c) + (-1 + d)x\right)}{\exp\left((2 + c) + (-1 + d)x\right) + \exp\left(c + dx\right)} \\ &= \frac{\exp\left(2 - x + c + dx\right)}{\exp\left(2 - x + c + dx\right) + \exp\left(c + dx\right)} \\ &= \frac{\exp(2 - x) \cdot \exp(c + dx)}{\exp(2 - x) \cdot \exp(c + dx) + 1 \cdot \exp(c + dx)} \\ &= \frac{\exp(2 - x) \cdot \exp(c + dx)}{\left(\exp(2 - x) + 1\right) \cdot \exp(c + dx)} \\ &= \frac{\exp(2 - x)}{1 + \exp(2 - x)} \end{aligned}$$

Since $\exp(c + dx)$ multiplies both the numerator and every term in the denominator, it factors out completely for **any** chosen values of $c$ and $d$.

**4. Why We Set $c = 0$ and $d = 0$ (Reference / Baseline Constraints)**

1. **Elimination of Over-Parameterization:**
In multi-class logistic regression / softmax models, the total number of parameters is redundant by 1 set of coefficients. By setting the baseline class (here, $\text{apple}$) parameters to zero:

$$c = 0 \implies \hat{\alpha}_{\text{apple}0} = 0$$


$$d = 0 \implies \hat{\alpha}_{\text{apple}1} = 0$$



we eliminate parameter indeterminacy and ensure unique coefficient estimation during model training (e.g., maximum likelihood estimation).
2. **Direct Interpretability:**
Setting $c = 0$ and $d = 0$ makes the remaining class parameters $(\hat{\alpha}_{\text{orange}0}, \hat{\alpha}_{\text{orange}1})$ directly equal to the standard logistic regression parameters $(\hat{\beta}_0, \hat{\beta}_1)$:

$$\hat{\alpha}_{\text{orange}0} = 2, \quad \hat{\alpha}_{\text{orange}1} = -1$$

(d)

* Intercept for orange: $\hat{\alpha}_{\text{orange}0} = 1.2$
* Slope for orange: $\hat{\alpha}_{\text{orange}1} = -2$
* Intercept for apple: $\hat{\alpha}_{\text{apple}0} = 3$
* Slope for apple: $\hat{\alpha}_{\text{apple}1} = 0.6$

The Intercept in my Model:

$$\hat{\beta}_0 = \hat{\alpha}_{\text{orange}0} - \hat{\alpha}_{\text{apple}0}$$

$$\hat{\beta}_0 = 1.2 - 3 = \mathbf{-1.8}$$

The Slope in my Model

$$\hat{\beta}_1 = \hat{\alpha}_{\text{orange}1} - \hat{\alpha}_{\text{apple}1}$$

$$\hat{\beta}_1 = -2 - 0.6 = \mathbf{-2.6}$$

That says: the coefficient estimates are:


$$\hat{\beta}_0 = -1.8 \quad \text{and} \quad \hat{\beta}_1 = -2.6$$

(e)

Standard binary logistic regression and two-class softmax regression are mathematically equivalent representations of the same model. Because both models produce the exact same estimated probability $\widehat{\operatorname{Pr}}(Y = \text{orange} \mid X = x)$ for any given value of $x$, their decision boundary (where $p(x) = 0.5$) is identical. Therefore, they will assign the exact same class label to every test observation, yielding an agreement fraction of **100%**
