```python
#(a)

x1 = np.linspace(-4, 2, 50)
x2 = np.linspace(-1, 5, 50)
X1, X2 = np.meshgrid(x1, x2)


Z = (1 + X1) ** 2 + (2 - X2) ** 2 - 4

fig, ax = plt.subplots(figsize=(6, 6))
contour = ax.contour(X1, X2, Z, levels=[0], colors="blue", linewidths=2)

ax.plot(-1, 2, "ro", label="Center (-1, 2)")

#(b)
## Shade the interior: (1 + X1)^2 + (2 - X2)^2 <= 4
ax.contourf(
    X1,
    X2,
    Z,
    levels=[-np.inf, 0],
    colors=["pink"],
    alpha=0.5,
    extend="lower",
)

## Shade the exterior: (1 + X1)^2 + (2 - X2)^2 > 4
ax.contourf(
    X1,
    X2,
    Z,
    levels=[0, np.inf],
    colors=["lightblue"],
    alpha=0.5,
    extend="upper",
)

ax.set_xlim(-4, 2)
ax.set_ylim(-1, 5)
ax.set_xlabel("$X_1$")
ax.set_ylabel("$X_2$")
ax.axhline(0, color="black", linestyle="--", lw=0.8)
ax.axvline(0, color="black", linestyle="--", lw=0.8)
ax.set_title("Sketching the Circle")
ax.legend(loc="upper right")
ax.grid(True, linestyle=":", alpha=0.6)
ax.set_aspect("equal")

plt.show()
```

The given decision boundary equation is:

$(1 + X_1)^2 + (2 - X_2)^2 = 4$


$(1 + X_1)^2 = 1 + 2X_1 + X_1^2$

$(2 - X_2)^2 = 4 - 4X_2 + X_2^2$

Substituting these back into the original equation:


$(1 + 2X_1 + X_1^2) + (4 - 4X_2 + X_2^2) = 4$

$X_1^2 + X_2^2 + 2X_1 - 4X_2 + 5 = 4$


$X_1^2 + X_2^2 + 2X_1 - 4X_2 + 1 = 0$

```python
#(a)
x1 = [3, 2, 4, 1, 2, 4, 4]
x2 = [4, 2, 4, 4, 1, 3, 1]
y = [1, 1, 1, 1, 0, 0, 0]

fig, ax = plt.subplots(figsize=(7, 6))
ax.scatter(x1, x2, c=y, cmap=plt.cm.coolwarm)

ax.set_xlabel("$X_1$")
ax.set_ylabel("$X_2$")


xx = np.linspace(0, 5, 50)

yy_optimal = xx - 0.5
ax.plot(
    xx,
    yy_optimal,
    "k-",
    linewidth=2,
    label=r"Optimal Hyperplane: $-0.5 + X_1 - X_2 = 0$",
)

#(d)
yy_margin1 = xx  # Passes through Class 1 support vectors (2,2) and (4,4)
yy_margin2 = xx - 1  # Passes through Class 0 support vectors (2,1) and (4,3)
ax.plot(xx, yy_margin1, "k--", alpha=0.6, label="Margin Boundary (Class 1)")
ax.plot(xx, yy_margin2, "k--", alpha=0.6, label="Margin Boundary (Class 0)")

ax.set_xlim(0, 5)
ax.set_ylim(0, 5)
ax.set_xlabel("$X_1$", fontsize=12)
ax.set_ylabel("$X_2$", fontsize=12)
ax.set_title("Maximal Margin Classifier", fontsize=14)
ax.legend(loc="upper left")
ax.grid(True, linestyle=":", alpha=0.6)
ax.set_aspect("equal")

plt.show()
```

(b)

$(2, 2)$ and $(4, 4)$ on Class 1:

$\text{Slope } m = \frac{4 - 2}{4 - 2} = 1$

$(2, 1)$ and $(4, 3)$ on Class 0:

$\text{Slope } m = \frac{3 - 1}{4 - 2} = 1$



$X_2 = 1 \cdot X_1 + b \quad \implies \quad X_2 - X_1 = b$



**For Class 1 boundary line** passing through $(2, 2)$ and $(4, 4)$:

$b_{\text{Class 1}} = 2 - 2 = 0 \quad \implies \quad X_2 = X_1$


**For Class 0 boundary line** passing through $(2, 1)$ and $(4, 3)$:

$b_{\text{Class 0}} = 1 - 2 = -1 \quad \implies \quad X_2 = X_1 - 1$





The maximal margin hyperplane lies **midway** between these two parallel boundaries:

$\text{Midpoint intercept } b = \frac{0 + (-1)}{2} = -0.5$

$X_2 = X_1 - 0.5$

$$-0.5 + X_1 - X_2 = 0$$

(c) 

Classification Rule and Coefficients ($\beta_0, \beta_1, \beta_2$)

*Rule:* Classify to *Red* ($y = 1$) if $-0.5 + X_1 - X_2 < 0$,and classify to *Blue* ($y = 0$) otherwise.

*Coefficients:*

$\beta_0 = -0.5$
$\beta_1 = 1$
$\beta_2 = -1$


(e) 


* Class 1 (Red): $(2, 2)$ and $(4, 4)$
* Class 0 (Blue): $(2, 1)$ and $(4, 3)$



(f)
Argument:

The maximal margin classifier depends only on the support vectors that lie on the margin boundaries. The seventh observation (at $(4, 1)$) lies safely outside the margin.
Therefore, a slight movement of the seventh observation will **not** affect the position of the optimal separating hyperplane, as long as it doesn't cross over the margin boundary or violate the classification boundary.


(g) 

* Any line that successfully separates the two classes without maximizing the margin is valid here.
* For example, we shift our line slightly so it still separates the classes but isn't centered. If we keep the slope of $1$ but change the intercept to $-1.8$:

$X_2 = X_1 - 1.8 \implies -1.8 + X_1 - X_2 = 0$


```python

plt.show()

x1 = [3, 2, 4, 1, 2, 4, 4]
x2 = [4, 2, 4, 4, 1, 3, 1]
y =  [1, 1, 1, 1, 0, 0, 0]

fig, ax = plt.subplots(figsize=(7, 6))
ax.scatter(x1, x2, c=y, cmap=plt.cm.coolwarm)

ax.set_xlabel("$X_1$")
ax.set_ylabel("$X_2$")


xx = np.linspace(0, 5, 50)

yy_non_optimal = xx - 1.8

ax.plot(
    xx,
    yy_non_optimal,
    "g--",
    label="Non-optimal Hyperplane: $-1.8 + X_1 - X_2 = 0$",
)
ax.set_xlim(0, 5)
ax.set_ylim(0, 5)
ax.set_xlabel("$X_1$", fontsize=12)
ax.set_ylabel("$X_2$", fontsize=12)
ax.set_title("Maximal Margin Classifier", fontsize=14)
ax.legend(loc="upper left")
ax.grid(True, linestyle=":", alpha=0.6)
ax.set_aspect("equal")

plt.show()
```

(h)
* To make the two classes no longer separable by a hyperplane, we need to add a point from one class right into the territory of the other class (overlapping them).
* For instance, if we add a red point ($y = 1$) right inside the blue class territory at $(1.5, 1.5)$, no straight line can ever cleanly separate red from blue anymore.


```python

x1 = [3, 2, 4, 1, 2, 4, 4]
x2 = [4, 2, 4, 4, 1, 3, 1]
y = [1, 1, 1, 1, 0, 0, 0]

fig, ax = plt.subplots(figsize=(7, 6))
ax.scatter(x1, x2, c=y, cmap=plt.cm.coolwarm)
#(h):

ax.scatter(1.5, 2, c=["Blue"],s=30, marker="+", label="Added Conflicting Obs")

ax.set_xlabel("$X_1$")
ax.set_ylabel("$X_2$")


xx = np.linspace(0, 5, 50)

yy_optimal = xx - 0.5
ax.plot(
    xx,
    yy_optimal,
    "k-",
    linewidth=2,
    label=r"Optimal Hyperplane: $-0.5 + X_1 - X_2 = 0$",
)


ax.set_xlim(0, 5)
ax.set_ylim(0, 5)
ax.set_xlabel("$X_1$", fontsize=12)
ax.set_ylabel("$X_2$", fontsize=12)
ax.set_title("Maximal Margin Classifier", fontsize=14)
ax.legend(loc="upper left")
ax.grid(True, linestyle=":", alpha=0.6)
ax.set_aspect("equal")

plt.show()
```

```python
import numpy as np
from matplotlib.pyplot import subplots, cm
import sklearn.model_selection as skm
from ISLP import load_data
from sklearn.svm import SVC
from ISLP.svm import plot as plot_svm
from sklearn.metrics import RocCurveDisplay
```

```python
rng = np.random.default_rng(1)
X = rng.standard_normal((100, 2))
X[:25] += 2
X[75:100] -= 2
y = np.array([1]*30+[2]*70)
```

```python
fig, ax = subplots(figsize=(8,8))
ax.scatter(X[:,0],X[:,1],c=y,cmap=cm.coolwarm)
```

```python
(X_train,X_test,y_train,y_test) = skm.train_test_split(X,
                                                       y,
                                                       test_size=0.5,
                                                       random_state=0)
svm_poly = SVC(kernel="poly", degree=3, C=1.0)
svm_poly.fit(X_train, y_train)
```

```python
svc_linear = SVC(kernel="linear", C=1.0)
svc_linear.fit(X_train, y_train)
```

```python
svm_rbf = SVC(kernel="rbf", gamma="scale", C=1.0)
svm_rbf.fit(X_train, y_train)
```

```python

def error_rate(model, X_data, y_data):
    return 1.0 - model.score(X_data, y_data)

print(
    f"Linear SVC  | Train Error: {error_rate(svc_linear, X_train, y_train):.2f} | Test Error: {error_rate(svc_linear, X_test, y_test):.2f}")
print(
    f"Poly Kernel | Train Error: {error_rate(svm_poly, X_train, y_train):.2f} | Test Error: {error_rate(svm_poly, X_test, y_test):.2f}")
print(
    f"RBF Kernel  | Train Error: {error_rate(svm_rbf, X_train, y_train):.2f} | Test Error: {error_rate(svm_rbf, X_test, y_test):.2f}")
```

```python
def plot_decision_boundary(model, ax, title):
    x_min, x_max = X[:, 0].min() - 1, X[:, 0].max() + 1
    y_min, y_max = X[:, 1].min() - 1, X[:, 1].max() + 1
    xx, yy = np.meshgrid(
        np.arange(x_min, x_max, 0.02), np.arange(y_min, y_max, 0.02)
    )

    Z = model.predict(np.c_[xx.ravel(), yy.ravel()])
    Z = Z.reshape(xx.shape)

   
    ax.contourf(xx, yy, Z, alpha=0.3, cmap=plt.cm.coolwarm)
    
    ax.scatter(
        X_train[:, 0],
        X_train[:, 1],
        c=y_train,
        cmap=plt.cm.coolwarm,
        edgecolors="k",
    )
    ax.set_title(title)


fig, axes = plt.subplots(1, 3, figsize=(15, 4))

plot_decision_boundary(
    svc_linear, axes[0], "Linear SVC (Flat Boundary)"
)
plot_decision_boundary(
    svm_poly, axes[1], "Polynomial Kernel (Degree 3)"
)
plot_decision_boundary(svm_rbf, axes[2], "Radial (RBF) Kernel")

plt.tight_layout()
plt.show()
```

The Radial (RBF) kernel typically performs best on test data because it creates smooth, flexible non-linear decision boundaries that fit non-linear clusters naturally without overfitting as easily as high-degree polynomials.

```python
#(f)
grid_rbf = GridSearchCV(
    SVC(kernel="rbf", gamma="scale"), c_grid, cv=5, scoring="accuracy"
)
grid_rbf.fit(X_train_scaled, y_train)

best_c_rbf = grid_rbf.best_params_["C"]
best_svc_rbf = SVC(kernel="rbf", C=best_c_rbf, gamma="scale")
best_svc_rbf.fit(X_train_scaled, y_train)

rbf_train_err = np.mean(best_svc_rbf.predict(X_train_scaled) != y_train)
rbf_test_err = np.mean(best_svc_rbf.predict(X_test_scaled) != y_test)

print(f"Radial SVM (Best C={best_c_rbf}):")
print(f"-> Support Vectors:     {len(best_svc_rbf.support_)}")
print(f"-> Training Error Rate: {rbf_train_err:.4f}")
print(f"-> Test Error Rate:     {rbf_test_err:.4f}")
```

```python
#(g)

grid_poly = GridSearchCV(
    SVC(kernel="poly", degree=2), c_grid, cv=5, scoring="accuracy"
)
grid_poly.fit(X_train_scaled, y_train)

best_c_poly = grid_poly.best_params_["C"]
best_svc_poly = SVC(kernel="poly", C=best_c_poly, degree=2)
best_svc_poly.fit(X_train_scaled, y_train)

poly_train_err = np.mean(best_svc_poly.predict(X_train_scaled) != y_train)
poly_test_err = np.mean(best_svc_poly.predict(X_test_scaled) != y_test)

print(f"Polynomial SVM (degree=2, Best C={best_c_poly}):")
print(f"-> Support Vectors:     {len(best_svc_poly.support_)}")
print(f"-> Training Error Rate: {poly_train_err:.4f}")
print(f"-> Test Error Rate:     {poly_test_err:.4f}")
```


The Radial Basis (RBF) kernel SVM achieves the best performance on the OJ dataset, yielding the lowest test error rate of 18.89%.

While the linear and radial models perform relatively similarly, the RBF kernel's slight flexibility allows it to capture subtle non-linear boundaries in consumer purchase behavior better than a strict linear plane or a degree-2 polynomial, which struggles with the highest test error here.
