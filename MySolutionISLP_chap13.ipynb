Exercise 1:

(a)

Suppose we test $m$ null hypotheses, all of which are true, and we control the Type I error for each test at level $\alpha$.

* **Derivation:** 
  A Type I error (false positive) occurs when we reject a true null hypothesis. By definition, the significance level $\alpha$ is the probability of making a Type I error on any single test:
  $$P(\text{Type I error for test } i) = \alpha$$

* **Expected Total:** 
  Using linearity of expectation, the expected number of false positives across all $m$ independent tests is simply the sum of individual expectations:
  $$\text{Expected Type I Errors} = \sum_{i=1}^{m} \mathbb{E}[\text{Indicator}_i] = m \times \alpha$$
 

(b)

* **The Complement Approach:** 
  Instead of calculating the probability of 1, 2, ..., $m$ errors directly, we look at the probability of getting **zero** errors (the complement):
  $$P(\text{zero false positives}) = \prod_{i=1}^{m} P(\text{test } i \text{ makes no error}) = (1 - \alpha)^m$$

* **Final Formula:**
  $$\text{FWER} = 1 - (1 - \alpha)^m$$
  * *Intuition:* As $m$ grows large, the chance of making *at least one* false discovery climbs rapidly toward 100%, illustrating why unadjusted multiple testing is problematic.


(c)

* **Qualitative Comparison:** 
  The FWER for positively correlated tests is **lower** than the FWER calculated under independence in part (b).

* **Intuition & Extreme Case:**
  * If the two tests are *perfectly correlated*, they act as a single test. The chance of making at least one false positive collapses down to just $\alpha$ ($0.05$), rather than expanding.
  * Because positive correlation causes tests to "cluster" or trigger together, their failure events overlap. This reduces the cumulative probability of getting at least one false positive compared to independent trials where errors are scattered randomly.


(d)

* **Intuition & Extreme Case:**
  * Suppose the tests are negatively correlated to the point where they are mutually exclusive (if one test triggers a false positive, the other is guaranteed not to). 
  * The probability of at least one false positive becomes the sum of their individual probabilities:
    $$\text{FWER} = P(\text{Test 1}) + P(\text{Test 2}) = \alpha + \alpha = 2\alpha$$
  * For $\alpha = 0.05$, this gives an upper bound FWER of $0.10$ (10%), which exceeds the independent case value of $2\alpha - \alpha^2 = 0.0975$ (9.75%). Negative correlation prevents error overlap, maximizing the spread of risk across the tests.

Exercise 2:

(a)

Let $A_j$ equal $1$ if the $j$-th null hypothesis is rejected (a Type I error), and $0$ otherwise.

* Since all null hypotheses are true, the probability of rejecting the $j$-th hypothesis is simply the significance level $\alpha$:
  $$P(A_j = 1) = \alpha, \quad P(A_j = 0) = 1 - \alpha$$
* Therefore, $A_j$ follows a **Bernoulli distribution** with success parameter $\alpha$:
  $$A_j \sim \text{Bernoulli}(\alpha)$$


(b) 

The sum $V = \sum_{j=1}^{m} A_j$ represents the total number of Type I errors made across all $m$ tests.

* Since all $m$ tests are independent, $V$ is the sum of $m$ independent and identically distributed (i.i.d.) Bernoulli($\alpha$) random variables.
* Therefore, $V$ follows a **Binomial distribution** with parameters $m$ and $\alpha$:
  $$\sum_{j=1}^{m} A_j \sim \text{Binomial}(m, \alpha)$$


(c)

Using the properties of the Binomial distribution:

* **Variance:** 
  $$\text{Var}\left(\sum_{j=1}^{m} A_j\right) = m \alpha (1 - \alpha)$$

* **Standard Deviation:**
  $$\text{SD}\left(\sum_{j=1}^{m} A_j\right) = \sqrt{m \alpha (1 - \alpha)}$$

* *Example:* If $m = 10,000$ and $\alpha = 0.05$:
  $$\text{Expected Errors} = 10,000 \times 0.05 = 500$$
  $$\text{Standard Deviation} = \sqrt{10,000 \times 0.05 \times 0.95} = \sqrt{475} \approx 21.79$$

Exercise 3:



1. **Define the Error Events:**
   Let $E_j$ be the event that a Type I error is committed on the $j$-th hypothesis (i.e., we falsely reject true null hypothesis $H_{0j}$). By design, we control each individual error rate such that:
   $$P(E_j) = \alpha_j$$

2. **Define FWER:**
   The Family-Wise Error Rate is the probability of making *at least one* Type I error across all $m$ tests:
   $$\text{FWER} = P\left(\bigcup_{j=1}^{m} E_j\right)$$

3. **Apply the Union Bound (Boole's Inequality):**
   According to Boole's inequality, the probability of the union of a collection of events is less than or equal to the sum of the probabilities of the individual events:
   $$P\left(\bigcup_{j=1}^{m} E_j\right) \le \sum_{j=1}^{m} P(E_j)$$

Substituting our individual error rates into the inequality:
   $$\text{FWER} \le \sum_{j=1}^{m} \alpha_j$$

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import statsmodels.api as sm
from ISLP import load_data
```

```python
from scipy.stats import \
(ttest_1samp,
ttest_rel,
ttest_ind,
t as t_dbn)
from statsmodels.stats.multicomp import \
pairwise_tukeyhsd
from statsmodels.stats.multitest import \
multipletests as mult_test
```

```python
Carseats = load_data('Carseats')
```

```python
Carseats.info()
```

```python
#(a)

quant_cols = ['CompPrice', 'Income', 'Advertising', 'Population', 'Price', 'Age', 'Education']

p_dict = {}
for col in quant_cols:
    X = sm.add_constant(Carseats[col])
    y = Carseats['Sales']
    model = sm.OLS(y, X).fit()
    p_dict[col] = model.pvalues[col]


df_p = pd.DataFrame(list(p_dict.items()), columns=['Variable', 'p_value'])
df_p = df_p.sort_values(by='p_value').reset_index(drop=True)

df_p
```

```python
#(b)
df_p['Reject_Unadjusted_0.05'] = df_p['p_value'] <= 0.05
```

```python

#(c)
reject_bonf, p_adj_bonf, _, _ = mult_test(df_p['p_value'], alpha=0.05, method='bonferroni')
df_p['Reject_FWER_Bonf_0.05'] = reject_bonf

#(d)

reject_fdr, p_adj_fdr, _, _ = mult_test(df_p['p_value'], alpha=0.20, method='fdr_bh')
df_p['Reject_FDR_BH_0.20'] = reject_fdr

df_p
```

```python
#(d)
rej_fdr_100, _, _, _ = mult_test(p_series, alpha=0.05, method='fdr_bh')
print(f"FDR BH Rejections: {rej_fdr_100.sum()}")
```

(e) Cherry-Picking the Top 10 Managers

If we select the 10 smallest $p$-values and pretend we only performed $m = 10$ tests:

**FWER Bonferroni ($\alpha = 0.05$ for $m = 10$):**
  * The adjusted threshold becomes $\frac{\alpha}{m_{\text{sub}}} = \frac{0.05}{10} = 0.005$.
  * Comparing the top 10 $p$-values against $0.005$ falsely yields **1 rejection** (since the smallest $p$-value in this simulation run falls below $0.005$).

**FDR Benjamini-Hochberg ($q = 0.05$ for $m = 10$):**
  * Evaluating the BH step-up thresholds with $m_{\text{sub}} = 10$ yields **1 rejection**.



(f) Why Cherry-Picking is Misleading

1. **Violation of the Search Space ($m=100$ vs $m=10$):**
   * Standard procedures like Bonferroni and Benjamini-Hochberg rely on the assumption that $m$ represents the **total number of hypotheses tested in the search process**.
   * By sifting through all 100 managers to pick the top 10, we implicitly evaluated all 100 tests. Setting $m = 10$ in the correction formulas ignores the 90 discarded tests, severely underestimating the true multiplicity of our experiment.

2. **Selection Bias & Inflation of Type I Errors:**
   * Under pure noise, order statistics guarantee that the minimum $p$-values out of 100 trials will naturally be very small.
   * Cherry-picking selects the extreme tails of random noise. Treating these pre-selected small $p$-values as if they were a random sample of 10 tests destroys Type I error control, leading to **false discoveries** even though no true signals exist.
