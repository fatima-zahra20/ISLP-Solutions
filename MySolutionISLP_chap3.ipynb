2. Carefully explain the differences between the KNN classifier and KNN
regression methods.

$$\text{Exercise 2: Answer :}$$Both methods work identically at the first step. Given a value of $K$ and a prediction point $x_0$, each identifies the $K$ training observations closest to $x_0$. They differ in what they do with those neighbours, because they differ in the type of response.

KNN regression is used when the response is quantitative. It estimates $f(x_0)$ as the average of the responses of the $K$ neighbours.

The KNN classifier is used when the response is qualitative. Rather than averaging, it estimates the conditional probability of each class as the fraction of the $K$ neighbours belonging to that class, then assigns $x_0$ to the class with the highest estimated probability. In effect it takes a majority vote among the neighbours.

Because the responses differ in type, performance is also measured differently: classification error rate for the classifier, and mean squared error for the regression method.

**$$\text{Exercise 5}$$**


**Start from the fitted value.** In regression without an intercept, the $i$th fitted value is $\hat{y}_i = x_i\hat\beta$, where

$$\hat\beta = \frac{\sum_{i'=1}^{n} x_{i'}y_{i'}}{\sum_{j=1}^{n} x_j^2}$$

The summation index inside $\hat\beta$ is written as $i'$ to distinguish it from the fixed observation $i$. The sum runs over all $n$ observations, whereas $x_i$ refers to one particular observation.

**Substitute.**

$$\hat{y}_i = x_i \cdot \frac{\sum_{i'=1}^{n} x_{i'}y_{i'}}{\sum_{j=1}^{n} x_j^2}$$

**Move $x_i$ inside the sum.** The term $x_i$ does not depend on the summation index $i'$, so it is constant with respect to that sum and may be brought inside it. The denominator is likewise constant with respect to $i'$:

$$\hat{y}_i = \frac{\sum_{i'=1}^{n} x_i x_{i'} y_{i'}}{\sum_{j=1}^{n} x_j^2} = \sum_{i'=1}^{n} \left( \frac{x_i x_{i'}}{\sum_{j=1}^{n} x_j^2} \right) y_{i'}$$

**Identify $a_{i'}$.** Comparing with the required form $\hat{y}_i = \sum_{i'=1}^{n} a_{i'} y_{i'}$:

$$a_{i'} = \frac{x_i \, x_{i'}}{\sum_{j=1}^{n} x_j^2}$$

**Interpretation.** No new quantity has been derived. The same fitted value has simply been rewritten, in the same way that $\frac{a+b+c}{d}$ can be rewritten as $\frac{a}{d} + \frac{b}{d} + \frac{c}{d}$. What the rewriting shows is that $\hat{y}_i$ is a linear combination of all $n$ observed responses, not of $y_i$ alone. The weight placed on $y_{i'}$ is proportional to $x_i x_{i'}$, so observations whose predictor values are large and of the same sign as $x_i$ contribute more heavily to the fitted value at $x_i$.

**$$\text{Exercise 6}$$**

least squares line is $\hat{y} = \hat\beta_0 + \hat\beta_1 x$. Substituting $x = \bar{x}$ and using the expression for $\hat\beta_0$ given in (3.4):

$$\hat{y} = \hat\beta_0 + \hat\beta_1\bar{x} = (\bar{y} - \hat\beta_1\bar{x}) + \hat\beta_1\bar{x} = \bar{y}$$

The terms in $\hat\beta_1\bar{x}$ cancel, so the fitted value at $x = \bar{x}$ is exactly $\bar{y}$. The least squares line therefore always passes through the point $(\bar{x}, \bar{y})$. This follows directly from the definition of $\hat\beta_0$, which is simply the equation $\bar{y} = \hat\beta_0 + \hat\beta_1\bar{x}$ rearranged.

**$$\text{Exercise 7}$$**

**Claim.** In simple linear regression of $Y$ onto $X$, the $R^2$ statistic equals the square of the correlation between $X$ and $Y$. We assume for simplicity that $\bar{x} = \bar{y} = 0$.

**Preliminaries.** When the sample means are zero, the least squares estimates in (3.4) simplify. The slope becomes $\hat\beta_1 = \left(\sum_{i=1}^n x_iy_i\right) / \left(\sum_{i=1}^n x_i^2\right)$, and the intercept $\hat\beta_0 = \bar{y} - \hat\beta_1\bar{x}$ reduces to zero. The fitted values are therefore $\hat{y}_i = \hat\beta_1 x_i$.

**The quantity to be matched.** From (3.18), the deviations $x_i - \bar{x}$ and $y_i - \bar{y}$ reduce to $x_i$ and $y_i$, so the correlation and its square are

$$\text{Cor}(X,Y) = \frac{\sum_{i=1}^n x_iy_i}{\sqrt{\sum_{i=1}^n x_i^2}\sqrt{\sum_{i=1}^n y_i^2}}, \qquad \text{Cor}(X,Y)^2 = \frac{\left(\sum_{i=1}^n x_iy_i\right)^2}{\left(\sum_{i=1}^n x_i^2\right)\left(\sum_{i=1}^n y_i^2\right)}$$

It remains to show that $R^2$ reduces to this same expression.

**Expanding the residual sum of squares.** Substituting the fitted values into the definition of RSS and expanding the square gives

$$\text{RSS} = \sum_{i=1}^n (y_i - \hat\beta_1 x_i)^2 = \sum_{i=1}^n y_i^2 - 2\hat\beta_1\sum_{i=1}^n x_iy_i + \hat\beta_1^2\sum_{i=1}^n x_i^2$$

Replacing $\hat\beta_1$ by its expression, the second term becomes $-2\left(\sum x_iy_i\right)^2 / \sum x_i^2$ and the third becomes $+\left(\sum x_iy_i\right)^2 / \sum x_i^2$. These share a common factor and combine, since $-2A + A = -A$, leaving

$$\text{RSS} = \sum_{i=1}^n y_i^2 - \frac{\left(\sum_{i=1}^n x_iy_i\right)^2}{\sum_{i=1}^n x_i^2}$$

**The total sum of squares.** Since $\bar{y} = 0$, the total sum of squares reduces to $\text{TSS} = \sum_{i=1}^n (y_i - \bar{y})^2 = \sum_{i=1}^n y_i^2$.

**Forming $R^2$.** Substituting both quantities into the definition $R^2 = 1 - \text{RSS}/\text{TSS}$, the terms $\sum y_i^2$ cancel in the numerator:

$$R^2 = \frac{\text{TSS} - \text{RSS}}{\text{TSS}} = \frac{\displaystyle\sum_{i=1}^n y_i^2 - \left(\sum_{i=1}^n y_i^2 - \frac{\left(\sum x_iy_i\right)^2}{\sum x_i^2}\right)}{\displaystyle\sum_{i=1}^n y_i^2} = \frac{\left(\sum_{i=1}^n x_iy_i\right)^2}{\left(\sum_{i=1}^n x_i^2\right)\left(\sum_{i=1}^n y_i^2\right)}$$

**Conclusion.** This is identical to the expression obtained for $\text{Cor}(X,Y)^2$, hence

$$R^2 = 1 - \frac{\text{RSS}}{\text{TSS}} = \text{Cor}(X,Y)^2 \qquad$$

The result explains why $R^2$ is interpreted as a measure of linear association in the simple regression setting: it carries exactly the same information as the correlation coefficient, differing only in that squaring discards the sign and so records the strength of the relationship without its direction.

**What the test compares.** The F-test compares two models. The first is a null model that ignores all predictors and simply predicts the mean value of mpg for every car. The second is the full model using all seven predictors. The test asks whether the full model reduces the residual sum of squares by more than would be expected by chance alone.

**The null hypothesis.** The F-test tests

$$H_0: \beta_1 = \beta_2 = \dots = \beta_7 = 0$$

The observed value of 252.4 is far larger. The p-value quantifies this: if all coefficients truly were zero, an F-statistic this large would occur essentially never.

**Conclusion.** We reject the null hypothesis and conclude that at least one of the predictors is associated with mpg. Note that the F-test establishes only that *some* relationship exists; it does not identify which predictors are responsible. That is determined from the individual t-tests in part (ii).

```python
#h) 

infl = model_e.get_influence()
stud = infl.resid_studentized_internal 
lev = infl.hat_matrix_diag

print("outliers:", (abs(stud) > 3).sum())
print("avg leverage:", 3/400) # p(predictors)+1 / number of observations
print("high leverage:", (lev > 3*(3/400)).sum())
```

*Outliers* are observations whose response value is far from the model's prediction. These are identified using studentized residuals, which express each residual as a number of standard errors. Values beyond $\pm 3$ are considered unusual.

*High leverage observations* are those with extreme predictor values, far from the centre of the other observations. These are identified using the leverage statistic $h_i$, whose average across all observations is always $(p+1)/n$. With $p = 2$ predictors and $n = 400$ observations, the average leverage here is $3/400 = 0.0075$. Values exceeding three times this average are flagged.
