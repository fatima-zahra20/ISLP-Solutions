Conceptual:

**Independent Censoring (Non-informative):** The mechanism causing a subject to be censored is **unrelated** to the event time / survival outcome (conditional on covariates). Examples include random loss-to-follow-up or fixed administrative cutoffs.
**Dependent Censoring (Informative):** The likelihood of being censored is **directly correlated** with the outcome of interest or underlying risk (e.g., sicker patients dropping out, or dropouts clustering among those with longer/shorter durations).





* (a) Phone numbers starting with "2" lost due to careless scientist

**Classification:** **Independent**;
**Reasoning:** Phone number prefixes are completely arbitrary and unrelated to disease relapse time.

* (b) Patient ages exceeding 99 lost due to a formatting error

**Classification:** **Independent**;
**Reasoning:** This is a fixed right-censoring rule ($T_C = 99$). The truncation threshold is administrative and does not depend on individual variations in underlying survival distributions once reaching age 99.

* (c) Very sick patients transferred to another hospital and lost

**Classification:** **Dependent**;
**Reasoning:** Patients closer to the event (death/shorter longevity) are systematically more likely to be lost to follow-up, correlating censoring with the survival outcome.

* (d) People who find work earlier are less motivated to stay in touch

**Classification:** **Dependent**;
**Reasoning:** Unemployment duration is the event time. Those lost to follow-up are specifically those experiencing the event faster (shorter unemployment duration).

* (e) Pre-term births more likely to occur away from hospital and be censored

**Classification:** **Dependent**;
**Reasoning:** Pre-term delivery corresponds directly to a shorter pregnancy duration, linking censoring probability to the event time.

* (f) Out-of-town college students lost to follow-up, and more likely to attend grad school

**Classification:** **Dependent**;
**Reasoning:** The outcome is total years of education. Students lost are systematically more likely to complete more years of education (attend graduate school), creating a correlation between censoring and educational attainment.

* (g) Patients not relapsing within 5 years considered cured and censored

**Classification:** **Independent**;
**Reasoning:** This represents fixed administrative right-censoring at a pre-determined time point ($C = 5\text{ years}$), applied uniformly regardless of individual risk profiles.

* (h) Iowa (opened 5 yrs ago) vs. Pittsburgh (opened 2 yrs ago) components; equal quality

**Classification:** **Independent**;
**Reasoning:** Since component quality/lifespan is identical between factories, the censoring times are purely determined by factory opening dates rather than component longevity.

* (i) Two factories opened at different times; older factory has higher quality

**Classification:** **Dependent**;
**Reasoning:** Component quality and longevity are explicitly tied to the factory, meaning component lifespan and censoring times are confounded/correlated.

Exercise 7:

(a)

**Argument under the Null Hypothesis (H0):**

1. **Null Hypothesis Statement:** Under H0, the survival distributions of Group 1 and Group 2 are identical. This implies that at any given event time y_(k), every individual in the combined risk set of size r_k has the exact same probability of experiencing an event, regardless of group membership.

2. **Conditional Sampling Structure:** At time y_(k), a total of q_k events occur out of the r_k available individuals at risk. Conditioned on the total event count q_k and total risk set size r_k, assigning r_1k of those individuals to Group 1 is equivalent to sampling a subset of size r_1k at random without replacement from a finite population of size r_k containing q_k success items.

3. **Hypergeometric Definition:** The distribution of the number of items of a specific class (q_1k events) obtained when drawing a fixed sample size (r_1k individuals) without replacement from a finite population (r_k individuals at risk) containing a fixed number of targeted items (q_k total events) is, by mathematical definition, a **Hypergeometric Distribution**.


**Parameter Mapping:**
A standard Hypergeometric distribution is parameterized as Hypergeometric(N, K, n):
- N (Total population size) = r_k
- K (Total number of success/failure items in population) = q_k
- n (Sample size extracted without replacement) = r_1k

Therefore: 
q_1k ~ Hypergeometric(N = r_k, K = q_k, n = r_1k)


(b)

**Expected Value E(q_1k)**

The general expectation formula for a hypergeometric random variable X ~ Hypergeometric(N, K, n) is:
E(X) = n * (K / N)

Substituting the parameters from our survival setup (N = r_k, K = q_k, n = r_1k):
E(q_1k) = r_1k * (q_k / r_k)
        = (r_1k / r_k) * q_k


The expected number of events in Group 1 is the proportion of Group 1 members in the risk set multiplied by the total number of events at that time point. This formally matches (11.5) in the text.




**Variance Var(q_1k)**

The general variance formula for a hypergeometric random variable X ~ Hypergeometric(N, K, n) is:
Var(X) = n * (K / N) * [1 - (K / N)] * [(N - n) / (N - 1)]

Substituting our parameters:
- n = r_1k
- K = q_k
- N = r_k
- N - n = r_k - r_1k = r_2k (since the individuals not at risk in Group 1 must be at risk in Group 2)

Var(q_1k) = r_1k * (q_k / r_k) * [1 - (q_k / r_k)] * [r_2k / (r_k - 1)]

Simplify the middle term [1 - (q_k / r_k)] into [(r_k - q_k) / r_k]:
Var(q_1k) = r_1k * (q_k / r_k) * [(r_k - q_k) / r_k] * [r_2k / (r_k - 1)]

Combining numerators and denominators:
Var(q_1k) = (q_k * r_1k * r_2k * (r_k - q_k)) / (r_k^2 * (r_k - 1))

This exact algebraic form accounts for finite population correction without replacement and formally matches (11.6) in the text.

Exercise 8:



1. **Survival Function S(t):** S(t) = P(T > t) = 1 - F(t), where F(t) = P(T ≤ t) is the cumulative distribution function (CDF).
   
2. **Probability Density Function f(t):** f(t) = dF(t)/dt = -dS(t)/dt.

3. **Hazard Function h(t):** The instantaneous failure rate at time t given survival up to time t:
   h(t) = lim_{delta_t -> 0} P(t ≤ T < t + delta_t | T ≥ t) / delta_t= f(t) / S(t)


**Proving f(t) = dF(t)/dt**


1. By definition, the cumulative distribution function is:
   F(t) = P(T ≤ t) = ∫_0^t f(u) du

2. Taking the derivative of F(t) with respect to t using the Fundamental Theorem of Calculus yields:
   dF(t)/dt = d/dt [ ∫_0^t f(u) du ] = f(t)

3. Alternatively, expressing F(t) in terms of S(t):
   F(t) = 1 - S(t)
   dF(t)/dt = d/dt [ 1 - S(t) ] = -dS(t)/dt

4. Since f(t) = -dS(t)/dt, we have:
   f(t) = dF(t)/dt


**Proving S(t) = exp( - ∫_0^t h(u) du )**


1. Start with the definition of the hazard rate:
   h(t) = f(t) / S(t)

2. Replace f(t) with -dS(t)/dt:
   h(t) = - [ dS(t)/dt ] / S(t)

3. Recognize that the right-hand side is the derivative of -ln(S(t)) via the chain rule:
   d/dt [ ln(S(t)) ] = (1 / S(t)) * (dS(t)/dt)
   
   Therefore:
   h(t) = - d/dt [ ln(S(t)) ]

4. Integrate both sides from 0 to t with respect to u:
   ∫_0^t h(u) du = - ∫_0^t (d/du [ ln(S(u)) ]) du
   ∫_0^t h(u) du = - [ ln(S(t)) - ln(S(0)) ]

5. Apply initial boundary condition at time t = 0:
   Everyone is alive at start, so S(0) = P(T > 0) = 1, and ln(S(0)) = ln(1) = 0.
   
   Thus:
   ∫_0^t h(u) du = - ln(S(t))

6. Multiply by -1 and exponentiate both sides:
   ln(S(t)) = - ∫_0^t h(u) du
   S(t) = exp( - ∫_0^t h(u) du )

Exercise 9:


(a): Show that S(t) = exp(-λt)

Using the Cumulative Distribution Function F(t)
1. The probability density function (PDF) is f(t) = λ exp(-λt) for t ≥ 0.
2. The cumulative distribution function F(t) is:
   F(t) = ∫_0^t f(u) du = ∫_0^t λ exp(-λu) du
        = [ -exp(-λu) ]_0^t
        = -exp(-λt) - (-exp(0))
        = 1 - exp(-λt)

3. Using the relationship S(t) = 1 - F(t) from Exercise 8:
   S(t) = 1 - (1 - exp(-λt)) = exp(-λt)

Using the Hazard Function h(t)
1. The hazard function is h(t) = f(t) / S(t).
   Since f(t) = λ exp(-λt) and S(t) = exp(-λt), we get:
   h(t) = (λ exp(-λt)) / exp(-λt) = λ  (constant hazard rate)

2. Using the integral relationship from Exercise 8:
   S(t) = exp( - ∫_0^t h(u) du )
        = exp( - ∫_0^t λ du )
        = exp(-λt)



(b): Likelihood Function L(λ)

For right-censored data with observed values y_i = min(t_i, c_i) and indicator δ_i:
- If an event occurs (δ_i = 1), the contribution to the likelihood is the density f(y_i).
- If censored (δ_i = 0), the contribution to the likelihood is the survival probability S(y_i).

The generic likelihood function (11.13) is:
L(λ) = ∏_{i=1}^n [ f(y_i) ]^{δ_i} [ S(y_i) ]^{1 - δ_i}

Substitute f(y_i) = λ exp(-λ y_i) and S(y_i) = exp(-λ y_i):
L(λ) = ∏_{i=1}^n [ λ exp(-λ y_i) ]^{δ_i} [ exp(-λ y_i) ]^{1 - δ_i}
     = ∏_{i=1}^n λ^{δ_i} exp(-λ y_i δ_i) exp(-λ y_i (1 - δ_i))
     = ∏_{i=1}^n λ^{δ_i} exp(-λ y_i [δ_i + 1 - δ_i])
     = ∏_{i=1}^n λ^{δ_i} exp(-λ y_i)

Taking the product across all n observations:
L(λ) = λ^(∑_{i=1}^n δ_i) * exp( -λ ∑_{i=1}^n y_i )



(c)

1. Take the log-likelihood function L(λ) = ln(L(λ)):
   L(λ) = ln( λ^(∑_{i=1}^n δ_i) * exp( -λ ∑_{i=1}^n y_i ) )
   L(λ) = (∑_{i=1}^n δ_i) ln(λ) - λ (∑_{i=1}^n y_i)

2. Take the first derivative with respect to λ:
   dL(λ)/dλ = (1 / λ) ∑_{i=1}^n δ_i - ∑_{i=1}^n y_i

3. Set the derivative equal to zero to solve for λ̂:
   (1 / λ̂) ∑_{i=1}^n δ_i - ∑_{i=1}^n y_i = 0
   (1 / λ̂) ∑_{i=1}^n δ_i = ∑_{i=1}^n y_i

4. Rearrange to isolate λ̂:
   λ̂ = ( ∑_{i=1}^n δ_i ) / ( ∑_{i=1}^n y_i )

(Interpretation: λ̂ = Total observed events / Total person-time at risk)




(d)

1. For an Exp(λ) random variable, the theoretical mean survival time E(T) is μ = 1 / λ.

2. By the plug-in property of Maximum Likelihood Estimators, the MLE estimator for the mean survival time μ̂ is:
   μ̂ = 1 / λ̂

3. Substituting our expression for λ̂ from part (c):
   μ̂ = ( ∑_{i=1}^n y_i ) / ( ∑_{i=1}^n δ_i )

(Interpretation: Estimated Mean Survival Time = Total person-time observed / Total number of events observed)

```python
df = pd.DataFrame(
    {
        'Y': [26.5, 37.2, 57.3, 90.8, 20.2, 89.8],
        'delta': [1, 1, 1, 0, 0, 0],
        'X': [0.1, 11.0, -0.3, 2.8, 1.8, 0.4],
    }
)

df['Group'] = (df['X'] >= 2).astype(int)

print(df)
```

```python

group1 = df[df['Group'] == 0]
group2 = df[df['Group'] == 1]


fig, ax = plt.subplots(figsize=(8, 5))


km1 = KaplanMeierFitter()
km1.fit(group1['Y'], event_observed=group1['delta'], label='Group 1 (X < 2)')
km1.plot_survival_function(ax=ax, color='blue', ci_show=False)

km2 = KaplanMeierFitter()
km2.fit(group2['Y'], event_observed=group2['delta'], label='Group 2 (X >= 2)')
km2.plot_survival_function(ax=ax, color='orange', ci_show=False)

plt.title('Kaplan-Meier Survival Curves by Group (Table 11.4)')
plt.xlabel('Time (Y)')
plt.ylabel('Estimated Survival Probability Ŝ(t)')
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend()
plt.show()
```

```python
#(b)
cph = CoxPHFitter()
cph.fit(df[['Y', 'delta', 'Group']], duration_col='Y', event_col='delta')

cph.print_summary()

beta_hat = cph.summary.loc['Group', 'coef']
hazard_ratio = cph.summary.loc['Group', 'exp(coef)']
p_value = cph.summary.loc['Group', 'p']

print(f'\nEstimated Coefficient (beta): {beta_hat:.4f}')
print(f'Hazard Ratio (exp(beta)): {hazard_ratio:.4f}')
print(f'p-value: {p_value:.4f}')
```

```python
#(c)
results = logrank_test(
    group1['Y'], group2['Y'], group1['delta'], group2['delta']
)

results.print_summary()

print(f'Log-Rank Test p-value: {results.p_value:.4f}')
print(f'Cox Model Score Test p-value: {cph.log_likelihood_ratio_test().p_value:.4f}')
```

The log-rank p-value (0.782) matches the Cox model score test p-value (0.78) almost exactly, illustrating their theoretical equivalence for a single binary predictor.
