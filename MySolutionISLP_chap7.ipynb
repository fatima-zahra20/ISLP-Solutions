When $x > \xi$, the truncated power basis element $(x - \xi)_+^3 = (x - \xi)^3$.

$(x - \xi)^3 = x^3 - 3\xi x^2 + 3\xi^2 x - \xi^3$

$f_2(x) = \beta_0 + \beta_1 x + \beta_2 x^2 + \beta_3 x^3 + \beta_4(x^3 - 3\xi x^2 + 3\xi^2 x - \xi^3)$

By collecting terms according to the powers of $x$, we get the cubic polynomial $f_2(x) = a_2 + b_2x + c_2x^2 + d_2x^3$ where:

* $a_2 = \beta_0 - \beta_4 \xi^3$ (Constant term)

(a): $\lambda = \infty, m = 0$

$\hat{g} = \arg\min_g \left( \sum_{i=1}^{n} (y_i - g(x_i))^2 + \lambda \int [g^{(m)}(x)]^2 dx \right)$

- $\lambda = \infty$ and $m = 0$

$\hat{g} = \arg\min_g \left( \sum_{i=1}^{n} (y_i - g(x_i))^2 + \infty \cdot \int [g(x)]^2 dx \right)$

for $\lambda = \infty$

$\int [g(x)]^2 dx = 0$


Since the integral of a squared real function can only be zero if the function itself is identically zero everywhere:


$$g(x) = 0$$

*(a):* The final equation is $g(x) = 0$ (a completely flat line lying on the horizontal axis).


(b): $\lambda = \infty, m = 1$

$\lambda = \infty$ and $m = 1$

$\hat{g} = \arg\min_g \left( \sum_{i=1}^{n} (y_i - g(x_i))^2 + \infty \cdot \int [g'(x)]^2 dx \right)$

for $\lambda = \infty$

To keep the objective function finite, the penalty term must be forced to zero:


$$\int [g'(x)]^2 dx = 0$$

For the integral of the squared first derivative (slope) to be zero, the first derivative must be zero everywhere:


$g'(x) = 0$

Integrating a slope of zero gives a constant value:

$g(x) = c$

*(b):* The final equation is a horizontal straight line ($g(x) = c$).


(c): $\lambda = \infty, m = 2$

$\lambda = \infty$ and $m = 2$

$\hat{g} = \arg\min_g \left( \sum_{i=1}^{n} (y_i - g(x_i))^2 + \infty \cdot \int [g''(x)]^2 dx \right)$

for $\lambda = \infty$

$\int [g''(x)]^2 dx = 0$

For the second derivative (curvature) to be zero everywhere:

$g''(x) = 0$


Integrating twice yields a linear function:

$g(x) = ax + b$

*(c):* The final equation is a **linear straight line** ($g(x) = ax + b$). *(Note: This is standard smoothing splines when $m=2$, which yields a cubic smoothing spline).*


(d): $\lambda = \infty, m = 3$

$\lambda = \infty$ and $m = 3$

$\hat{g} = \arg\min_g \left( \sum_{i=1}^{n} (y_i - g(x_i))^2 + \infty \cdot \int [g'''(x)]^2 dx \right)$

for $\lambda = \infty$

$\int [g'''(x)]^2 dx = 0$


For the third derivative to be zero everywhere:

$g'''(x) = 0$

Integrating three times yields a quadratic polynomial:

$g(x) = ax^2 + bx + c$

*(d):* The final equation is a **quadratic curve / parabola** ($g(x) = ax^2 + bx + c$).


(e): $\lambda = 0, m = 3$

$\lambda = 0$ and $m = 3$

$\hat{g} = \arg\min_g \left( \sum_{i=1}^{n} (y_i - g(x_i))^2 + 0 \cdot \int [g'''(x)]^2 dx \right)$


Since $\lambda = 0$, the entire roughness penalty term is multiplied by zero and wiped out completely:


$\hat{g} = \arg\min_g \left( \sum_{i=1}^{n} (y_i - g(x_i))^2 \right)$

With zero penalty for roughness, the formula cares exclusively about minimizing training error to zero:

$g(x_i) = y_i \quad \text{for all } i$

*Conclusion for (e):* The final equation is an **interpolating curve**—a flexible curve that twists and bends to pass directly through every single data point $(x_i, y_i)$.

Exercise 5:

(a) As $\lambda \to \infty$, will $\hat{g}_1$ or $\hat{g}_2$ have the smaller training RSS?

**Answer:** **$\hat{g}_2$ will have the smaller training RSS.**

**Because:**
when $\lambda \to \infty$, the penalty forces the $m$-th derivative to be zero:
* For $\hat{g}_1$ ($m = 3$), the 3rd derivative is zero, which forces it to be a *quadratic curve* ($ax^2 + bx + c$).

* For $\hat{g}_2$ ($m = 4$), the 4th derivative is zero, which forces it to be a **cubic polynomial** ($ax^3 + bx^2 + cx + d$).


(b) As $\lambda \to \infty$, will $\hat{g}_1$ or $\hat{g}_2$ have the smaller test RSS?

**Answer:** It depends on the true underlying relationship, but typically $\hat{g}_1$ (or they could be very close), though usually lower flexibility wins out at extreme penalties.

**Beacuse**
* When $\lambda \to \infty$, both models are heavily constrained. $\hat{g}_1$ is a quadratic curve and $\hat{g}_2$ is a cubic curve.

* If the true relationship is linear or quadratic, $\hat{g}_1$ avoids unnecessary variance and might win on test data. However, if the true function has slight cubic curvature, $\hat{g}_2$ might capture it better. Generally speaking, both suffer from high bias when $\lambda \to \infty$, but $\hat{g}_1$ is simpler.


(c) For $\lambda = 0$, will $\hat{g}_1$ or $\hat{g}_2$ have the smaller training and test RSS?

**Answer:** 
*Training RSS:* They will be equal (both equal to 0).

*Test RSS:* Neither has a guaranteed advantage (both will likely overfit badly).

**Because**
When $\lambda = 0$, the penalty term disappears completely for *both* curves.
Both $\hat{g}_1$ and $\hat{g}_2$ reduce to ordinary curve interpolation where they bend and twist to pass **directly through every single training point** ($g(x_i) = y_i$).

```python
#(a)
np.random.seed(0)
n = 100
X1 = np.random.normal(size=n)
X2 = np.random.normal(size=n)

Y = 2.5 + 1.5 * X1 - 2.0 * X2 + np.random.normal(size=n)
```

```python
# (b)
def simple_reg(outcome, feature):
  X = sm.add_constant(feature)
  model = sm.OLS(outcome, X).fit()
  return model.params[0], model.params[1]


# (c) 
beta1 = 2

# (d)

outcome_d = Y - beta1 * X1
beta0, beta2 = simple_reg(outcome_d, X2)
print('beta0:', beta0, 'beta2:', beta2)
```

```python

# (e)
outcome_e = Y - beta2 * X2
beta0, beta1 = simple_reg(outcome_e, X1)
print('beta0:',beta0, 'beta1:' ,beta1 ,'beta2', beta2)
```

```python

num_iters = 1000

beta0_history = np.zeros(num_iters)
beta1_history = np.zeros(num_iters)
beta2_history = np.zeros(num_iters)


beta1 = 2

for i in range(num_iters):
  beta0, beta2 = simple_reg(Y - beta1 * X1, X2)
  beta0, beta1 = simple_reg(Y - beta2 * X2, X1)

 
  beta0_history[i] = beta0
  beta1_history[i] = beta1
  beta2_history[i] = beta2


fig, ax = plt.subplots(figsize=(10, 5))
ax.plot(range(num_iters), beta0_history, label="beta0", color="blue")
ax.plot(range(num_iters), beta1_history, label="beta1", color="green")
ax.plot(range(num_iters), beta2_history, label="beta2", color="red")
ax.set_title("Backfitting Convergence over 1,000 Iterations")
ax.set_xlabel("Iteration")
ax.set_ylabel("Coefficient Estimate")
ax.legend()
ax.grid(True, alpha=0.3)

plt.show()
```

3 to 4 iteration are actually enough!

```python

np.random.seed(0)
n = 100
p = 100

X = np.random.normal(size=(n, p))
true_beta = np.random.normal(size=p + 1)
Y = true_beta[0] + np.sum(X * true_beta[1:], axis=1) + np.random.normal(size=n)


beta = np.zeros(p + 1)
num_iters = 15
beta_history = np.zeros((num_iters, p + 1))


for it in range(num_iters):
  pred_sum = np.sum(X * beta[1:], axis=1)
  beta[0] = np.mean(Y - pred_sum)

  for j in range(p):
    partial_resid = Y - beta[0]
    for k in range(p):
      if k != j:
        partial_resid = partial_resid - (X[:, k] * beta[k + 1])

    _, beta[j + 1] = simple_reg(partial_resid, X[:, j])

  beta_history[it, :] = beta

X_mult = sm.add_constant(X)
true_coefs = sm.OLS(Y, X_mult).fit().params

plt.figure(figsize=(10, 5))
for j in range(p + 1):
  plt.plot(range(num_iters), beta_history[:, j], color="gray", alpha=0.3)

plt.title("Backfitting Convergence (p = 100 Predictors)")
plt.xlabel("Iteration")
plt.ylabel("Coefficient Estimates")
plt.grid(True, alpha=0.3)
plt.show()
```

A "good" approximation is effectively the exact least-squares solution within just 3 to 4 iterations.

the coordinate descent updates stabilize almost immediately after a few full cycles through all predictors, proving that backfitting scales efficiently even in higher-dimensional settings ($p = 100$).
