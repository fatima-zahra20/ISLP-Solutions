Exercise 1:

(a)

$f(x) = \beta_0 + \beta_1x + \beta_2x^2 + \beta_3x^3 + \beta_4(x - \xi)_+^3$ eliminating the last term, leaving:

$f_1(x) = \beta_0 + \beta_1 x + \beta_2 x^2 + \beta_3 x^3$

Thus, the coefficients expressed in terms of $\beta$ are:

* $a_1 = \beta_0$
* $b_1 = \beta_1$
* $c_1 = \beta_2$
* $d_1 = \beta_3$



(b)

When $x > \xi$, the truncated power basis element $(x - \xi)_+^3 = (x - \xi)^3$.

$(x - \xi)^3 = x^3 - 3\xi x^2 + 3\xi^2 x - \xi^3$

$f_2(x) = \beta_0 + \beta_1 x + \beta_2 x^2 + \beta_3 x^3 + \beta_4(x^3 - 3\xi x^2 + 3\xi^2 x - \xi^3)$

By collecting terms according to the powers of $x$, we get the cubic polynomial $f_2(x) = a_2 + b_2x + c_2x^2 + d_2x^3$ where:

* $a_2 = \beta_0 - \beta_4 \xi^3$ (Constant term)
* $b_2 = \beta_1 + 3\beta_4 \xi^2$ (Linear coefficient)
* $c_2 = \beta_2 - 3\beta_4 \xi$ (Quadratic coefficient)
* $d_2 = \beta_3 + \beta_4$ (Cubic coefficient)

(c)


- Evaluate $f_1(\xi)$ (using our polynomial for $x \le \xi$ from Part a):

$f_1(\xi) = \beta_0 + \beta_1\xi + \beta_2\xi^2 + \beta_3\xi^3$


- Evaluate $f_2(\xi)$ (using our expanded polynomial for $x > \xi$ from Part b, substituting our expressions for $a_2, b_2, c_2, d_2$):

$f_2(\xi) = (\beta_0 - \beta_4\xi^3) + (\beta_1 + 3\beta_4\xi^2)\xi + (\beta_2 - 3\beta_4\xi)\xi^2 + (\beta_3 + \beta_4)\xi^3$


- Expand and simplify $f_2(\xi)$:

$f_2(\xi) = \beta_0 - \beta_4\xi^3 + \beta_1\xi + 3\beta_4\xi^3 + \beta_2\xi^2 - 3\beta_4\xi^3 + \beta_3\xi^3 + \beta_4\xi^3$


all the $\beta_4\xi^3$ terms cancel each other out ($-\beta_4\xi^3 + 3\beta_4\xi^3 - 3\beta_4\xi^3 + \beta_4\xi^3 = 0$):

$f_2(\xi) = \beta_0 + \beta_1\xi + \beta_2\xi^2 + \beta_3\xi^3$



Since $f_1(\xi) = f_2(\xi)$, *the function is continuous at $\xi$*.



(d)


Using the derivative formula $f'(x) = b + 2cx + 3dx^2$:

- For $f_1(x)$ (where $b_1 = \beta_1, c_1 = \beta_2, d_1 = \beta_3$):

$f_1'(\xi) = \beta_1 + 2\beta_2\xi + 3\beta_3\xi^2$


- For $f_2(x)$** (using our coefficients $b_2, c_2, d_2$ from Part b):

$f_2'(\xi) = (\beta_1 + 3\beta_4\xi^2) + 2(\beta_2 - 3\beta_4\xi)\xi + 3(\beta_3 + \beta_4)\xi^2$


- Expand and simplify $f_2'(\xi)$:

$f_2'(\xi) = \beta_1 + 3\beta_4\xi^2 + 2\beta_2\xi - 6\beta_4\xi^2 + 3\beta_3\xi^2 + 3\beta_4\xi^2$

$f_2'(\xi) = \beta_1 + 2\beta_2\xi + 3\beta_3\xi^2$



Since $f_1'(\xi) = f_2'(\xi)$, *the first derivative is continuous at $\xi$*



(e)

Using the second derivative formula $f''(x) = 2c + 6dx$:

- For $f_1(x)$ (where $c_1 = \beta_2, d_1 = \beta_3$):

$f_1''(\xi) = 2\beta_2 + 6\beta_3\xi$


- For $f_2(x)$ (using our coefficients $c_2, d_2$ from Part b):

$f_2''(\xi) = 2(\beta_2 - 3\beta_4\xi) + 6(\beta_3 + \beta_4)\xi$


- Expand and simplify $f_2''(\xi)$:

$f_2''(\xi) = 2\beta_2 - 6\beta_4\xi + 6\beta_3\xi + 6\beta_4\xi$


Notice how the $-6\beta_4\xi$ and $+6\beta_4\xi$ cancel out to $0$:

$f_2''(\xi) = 2\beta_2 + 6\beta_3\xi$



Since $f_1''(\xi) = f_2''(\xi)$, *the second derivative is continuous at $\xi$*


Because $f(x)$, $f'(x)$, and $f''(x)$ are all continuous at the knot $\xi$, $f(x)$ satisfies the formal definition of a **cubic regression spline**.

Exercise 2:

(a): $\lambda = \infty, m = 0$

$\hat{g} = \arg\min_g \left( \sum_{i=1}^{n} (y_i - g(x_i))^2 + \lambda \int [g^{(m)}(x)]^2 dx \right)$

- $\lambda = \infty$ and $m = 0$

$\hat{g} = \arg\min_g \left( \sum_{i=1}^{n} (y_i - g(x_i))^2 + \infty \cdot \int [g(x)]^2 dx \right)$

for $\lambda = \infty$

$\int [g(x)]^2 dx = 0$


Since the integral of a squared real function can only be zero if the function itself is identically zero everywhere:


$$g(x) = 0$$

*(a):* The final equation is $g(x) = 0$ (a completely flat line lying on the horizontal axis).


(b): $\lambda = \infty, m = 1$

$\lambda = \infty$ and $m = 1$

$\hat{g} = \arg\min_g \left( \sum_{i=1}^{n} (y_i - g(x_i))^2 + \infty \cdot \int [g'(x)]^2 dx \right)$

for $\lambda = \infty$

To keep the objective function finite, the penalty term must be forced to zero:


$$\int [g'(x)]^2 dx = 0$$

For the integral of the squared first derivative (slope) to be zero, the first derivative must be zero everywhere:


$g'(x) = 0$

Integrating a slope of zero gives a constant value:

$g(x) = c$

*(b):* The final equation is a horizontal straight line ($g(x) = c$).


(c): $\lambda = \infty, m = 2$

$\lambda = \infty$ and $m = 2$

$\hat{g} = \arg\min_g \left( \sum_{i=1}^{n} (y_i - g(x_i))^2 + \infty \cdot \int [g''(x)]^2 dx \right)$

for $\lambda = \infty$

$\int [g''(x)]^2 dx = 0$

For the second derivative (curvature) to be zero everywhere:

$g''(x) = 0$


Integrating twice yields a linear function:

$g(x) = ax + b$

*(c):* The final equation is a **linear straight line** ($g(x) = ax + b$). *(Note: This is standard smoothing splines when $m=2$, which yields a cubic smoothing spline).*


(d): $\lambda = \infty, m = 3$

$\lambda = \infty$ and $m = 3$

$\hat{g} = \arg\min_g \left( \sum_{i=1}^{n} (y_i - g(x_i))^2 + \infty \cdot \int [g'''(x)]^2 dx \right)$

for $\lambda = \infty$

$\int [g'''(x)]^2 dx = 0$


For the third derivative to be zero everywhere:

$g'''(x) = 0$

Integrating three times yields a quadratic polynomial:

$g(x) = ax^2 + bx + c$

*(d):* The final equation is a **quadratic curve / parabola** ($g(x) = ax^2 + bx + c$).


(e): $\lambda = 0, m = 3$

$\lambda = 0$ and $m = 3$

$\hat{g} = \arg\min_g \left( \sum_{i=1}^{n} (y_i - g(x_i))^2 + 0 \cdot \int [g'''(x)]^2 dx \right)$


Since $\lambda = 0$, the entire roughness penalty term is multiplied by zero and wiped out completely:


$\hat{g} = \arg\min_g \left( \sum_{i=1}^{n} (y_i - g(x_i))^2 \right)$

With zero penalty for roughness, the formula cares exclusively about minimizing training error to zero:

$g(x_i) = y_i \quad \text{for all } i$

*Conclusion for (e):* The final equation is an **interpolating curve**—a flexible curve that twists and bends to pass directly through every single data point $(x_i, y_i)$.

$\hat{Y} = 1 + 1 \cdot (X) + (-2) \cdot (X - 1)^2 I(X \ge 1)$

When $X < 1$

Because $X < 1$, the indicator function $I(X \ge 1) = 0$. The second term disappears entirely:

$\hat{Y} = 1 + X$

A straight line with an intercept of $1$ and a slope of $1$.

$\hat{Y} = 1 + X - 2(X - 1)^2$

$\hat{Y} = 1 + X - 2(X^2 - 2X + 1)$

$\hat{Y} = 1 + X - 2X^2 + 4X - 2$

$\hat{Y} = -1 + 5X - 2X^2$

A downward-opening parabola (quadratic curve) for values where $X \ge 1$.

Exercise 4:

$\hat{Y} = \hat{\beta}_0 + \hat{\beta}_1 b_1(X) + \hat{\beta}_2 b_2(X)$

$\hat{Y} = 1 + 1 \cdot \big[ I(0 \le X \le 2) - (X-1)I(1 \le X \le 2) \big] + 3 \cdot \big[ (X-3)I(3 \le X \le 4) + I(4 < X \le 5) \big]$


**Interval 1: $X \in [-2, 0)$**

* Both indicator blocks are $0$.
* $\hat{Y} = 1 + 0 + 0 = \mathbf{1}$
* **Shape:** A flat horizontal line at $Y = 1$.


**Interval 2: $X \in [0, 1)$**

* $I(0 \le X \le 2) = 1$, but $I(1 \le X \le 2) = 0$.
* $\hat{Y} = 1 + 1(1 - 0) + 0 = \mathbf{2}$
* **Shape:** A step up to a flat horizontal line at $Y = 2$.


**Interval 3: $X \in [1, 2]$**

* Both parts of $b_1$ are active: $I(0 \le X \le 2) = 1$ and $I(1 \le X \le 2) = 1$.
* $\hat{Y} = 1 + 1 \cdot (1 - (X - 1)) + 0 = 1 + 1 - X + 1 = \mathbf{3 - X}$
* **Shape:** A downward sloping line. At $X = 1$, $\hat{Y} = 2$. At $X = 2$, $\hat{Y} = 1$.


**Interval 4: $X \in (2, 3)$**

* All indicators are $0$ here (since $b_1$ only goes up to $2$ and $b_2$ starts at $3$).
* $\hat{Y} = 1 + 0 + 0 = \mathbf{1}$
* **Shape:** Drops down to a flat line at $Y = 1$.


**Interval 5: $X \in [3, 4]$**

* The first part of $b_2$ is active: $I(3 \le X \le 4) = 1$.
* $\hat{Y} = 1 + 0 + 3(X - 3) = 1 + 3X - 9 = \mathbf{3X - 8}$
* **Shape:** An upward sloping line. At $X = 3$, $\hat{Y} = 1$. At $X = 4$, $\hat{Y} = 4$.


**Interval 6: $X \in (4, 5]$**

* The second part of $b_2$ is active: $I(4 < X \le 5) = 1$.
* $\hat{Y} = 1 + 0 + 3(1) = \mathbf{4}$
* **Shape:** A flat horizontal line at $Y = 4$.


**Interval 7: $X \in (5, 6]$**

* All indicators are $0$ again.
* $\hat{Y} = 1 + 0 + 0 = \mathbf{1}$
* **Shape:** Drops back down to a flat line at $Y = 1$.


* **From $X = -2$ to $0$:** Flat line at $Y = 1$.
* **From $X = 0$ to $1$:** Flat line at $Y = 2$.
* **From $X = 1$ to $2$:** Slanted line connecting $(1, 2)$ down to $(2, 1)$.
* **From $X = 2$ to $3$:** Flat line at $Y = 1$.
* **From $X = 3$ to $4$:** Slanted line connecting $(3, 1)$ up to $(4, 4)$.
* **From $X = 4$ to $5$:** Flat line at $Y = 4$.
* **From $X = 5$ to $6$:** Flat line dropping back down to $Y = 1$.

Exercise 5:

(a) As $\lambda \to \infty$, will $\hat{g}_1$ or $\hat{g}_2$ have the smaller training RSS?

**Answer:** **$\hat{g}_2$ will have the smaller training RSS.**

**Because:**
when $\lambda \to \infty$, the penalty forces the $m$-th derivative to be zero:
* For $\hat{g}_1$ ($m = 3$), the 3rd derivative is zero, which forces it to be a *quadratic curve* ($ax^2 + bx + c$).

* For $\hat{g}_2$ ($m = 4$), the 4th derivative is zero, which forces it to be a **cubic polynomial** ($ax^3 + bx^2 + cx + d$).


(b) As $\lambda \to \infty$, will $\hat{g}_1$ or $\hat{g}_2$ have the smaller test RSS?

**Answer:** It depends on the true underlying relationship, but typically $\hat{g}_1$ (or they could be very close), though usually lower flexibility wins out at extreme penalties.

**Beacuse**
* When $\lambda \to \infty$, both models are heavily constrained. $\hat{g}_1$ is a quadratic curve and $\hat{g}_2$ is a cubic curve.

* If the true relationship is linear or quadratic, $\hat{g}_1$ avoids unnecessary variance and might win on test data. However, if the true function has slight cubic curvature, $\hat{g}_2$ might capture it better. Generally speaking, both suffer from high bias when $\lambda \to \infty$, but $\hat{g}_1$ is simpler.


(c) For $\lambda = 0$, will $\hat{g}_1$ or $\hat{g}_2$ have the smaller training and test RSS?

**Answer:** 
*Training RSS:* They will be equal (both equal to 0).

*Test RSS:* Neither has a guaranteed advantage (both will likely overfit badly).

**Because**
When $\lambda = 0$, the penalty term disappears completely for *both* curves.
Both $\hat{g}_1$ and $\hat{g}_2$ reduce to ordinary curve interpolation where they bend and twist to pass **directly through every single training point** ($g(x_i) = y_i$).
