1. Describe the null hypotheses to which the p-values given in Table 3.4
correspond. Explain what conclusions you can draw based on these
p-values. Your explanation should be phrased in terms of sales, TV,
radio, and newspaper, rather than in terms of the coefficients of the
linear model.

TABLE 3.4. For the Advertising data, least squares coefficient estimates of the
multiple linear regression of number of units sold on TV, radio, and newspaper
advertising budgets. page 83

$$\text{Exercise 1: Answer :}$$Each p-value tests its own null hypothesis: that the predictor in that row has no effect on sales once the other two predictors are already accounted for.

For TV and radio the p-values are very small, so we reject those null hypotheses. Knowing the TV budget tells us something about sales that radio and newspaper do not, and the same is true of radio. Both are genuinely associated with sales.

For newspaper the p-value is 0.8599, which is large, so we have no evidence against its null hypothesis. Once we already know the TV and radio budgets, the newspaper budget adds nothing further about sales.

This is worth noting because a simple regression of sales on newspaper alone does suggest a relationship. The reason is that newspaper and radio budgets tend to rise and fall together. With radio missing from the model, newspaper stands in for it and looks effective. Once radio is included, that apparent effect disappears.

2. Carefully explain the differences between the KNN classifier and KNN
regression methods.

$$\text{Exercise 2: Answer :}$$Both methods work identically at the first step. Given a value of $K$ and a prediction point $x_0$, each identifies the $K$ training observations closest to $x_0$. They differ in what they do with those neighbours, because they differ in the type of response.

KNN regression is used when the response is quantitative. It estimates $f(x_0)$ as the average of the responses of the $K$ neighbours.

The KNN classifier is used when the response is qualitative. Rather than averaging, it estimates the conditional probability of each class as the fraction of the $K$ neighbours belonging to that class, then assigns $x_0$ to the class with the highest estimated probability. In effect it takes a majority vote among the neighbours.

Because the responses differ in type, performance is also measured differently: classification error rate for the classifier, and mean squared error for the regression method.

**$$\text{Exercise 3(a): Answer iii}$$**

**The model.** With $\hat\beta_0 = 50$, $\hat\beta_1 = 20$, $\hat\beta_2 = 0.07$, $\hat\beta_3 = 35$, $\hat\beta_4 = 0.01$, $\hat\beta_5 = -10$:

$$\hat{y} = 50 + 20\,\text{GPA} + 0.07\,\text{IQ} + 35\,\text{Level} + 0.01(\text{GPA} \times \text{IQ}) - 10(\text{GPA} \times \text{Level})$$

**Write the equation for each group.** Substituting $\text{Level} = 0$ for high school graduates:

$$\hat{y}_{HS} = 50 + 20\,\text{GPA} + 0.07\,\text{IQ} + 0.01(\text{GPA} \times \text{IQ})$$

Substituting $\text{Level} = 1$ for college graduates:

$$\hat{y}_{C} = 50 + 20\,\text{GPA} + 0.07\,\text{IQ} + 0.01(\text{GPA} \times \text{IQ}) + 35 - 10\,\text{GPA}$$

**Subtract to find the difference.** The college equation contains every term of the high school equation plus two additional terms, so all shared terms cancel:

$$\hat{y}_{C} - \hat{y}_{HS} = 35 - 10\,\text{GPA}$$

The comparison depends only on GPA. IQ appears in the model but affects both groups identically, so it plays no role in the difference.

**Evaluate the difference.**

| GPA | $35 - 10\,\text{GPA}$ | Interpretation |
|---|---|---|
| 2.0 | $+15$ | College graduates earn \$15k more |
| 3.0 | $+5$ | College graduates earn \$5k more |
| 3.5 | $0$ | The two groups earn the same |
| 4.0 | $-5$ | High school graduates earn \$5k more |

Setting the difference to zero gives the crossover point:

$$35 - 10\,\text{GPA} = 0 \implies \text{GPA} = 3.5$$

**$$\text{Exercise 3(b)}$$**

Predict the salary of a college graduate with IQ of 110 and GPA of 4.0.

Substituting $\text{GPA} = 4.0$, $\text{IQ} = 110$ and $\text{Level} = 1$ into the fitted model:

$$\hat{y} = 50 + 20(4.0) + 0.07(110) + 35(1) + 0.01(4.0 \times 110) - 10(4.0 \times 1)$$

Evaluating each term:

$$\hat{y} = 50 + 80 + 7.7 + 35 + 4.4 - 40 = 137.1$$

Since the response is measured in thousands of dollars, the predicted starting salary is **\$137,100**.

Note that the two interaction terms require multiplying the variables together before applying the coefficient: $\text{GPA} \times \text{IQ} = 440$, and $\text{GPA} \times \text{Level} = 4.0$. The error term is not included, since a prediction uses the fitted equation only.


**$$\text{Exercise 3(c)}$$**
False. A small coefficient does not mean a small effect, because the size of a coefficient depends on the scale of the variable it multiplies. The interaction term here is $\text{GPA} \times \text{IQ}$, which takes values in the hundreds. For a student with a GPA of 4.0 and an IQ of 110, the product is 440, so the term contributes $0.01 \times 440 = 4.4$, or \$4,400 to the predicted salary. That is not a negligible amount.
More importantly, the size of a coefficient does not tell us whether an effect is real. That comes from the p-value, which compares the coefficient to its standard error. We are given neither for this term, so we cannot conclude that the evidence for an interaction is weak.
